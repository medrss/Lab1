## Задание 2

Построить график функции:

\[
y(x)=\frac{x+\cos(x+\pi)}{x^2}
\]

В точке `x = 0` функция не определена, поэтому около нуля используется разрыв.

```python
# Задание 2. График функции

x_left = np.linspace(-10, -0.05, 1000)
x_right = np.linspace(0.05, 10, 1000)

y_left = (x_left + np.cos(x_left + np.pi)) / (x_left ** 2)
y_right = (x_right + np.cos(x_right + np.pi)) / (x_right ** 2)

plt.figure(figsize=(10, 6))
plt.plot(x_left, y_left, label="y(x)")
plt.plot(x_right, y_right)

plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1, linestyle="--", label="x = 0")

plt.title("График функции y(x) = (x + cos(x + π)) / x²")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

## Задание 3

Сгенерировать **1000 точек** `(x, y)`. Каждая координата имеет нормальное распределение с:
- математическим ожиданием `μ = 0.1`;
- дисперсией `D = 5`.

Стандартное отклонение равно:

\[
\sigma=\sqrt{5}
\]

Ниже строятся диаграмма рассеяния и тепловая карта плотности точек.

```python
# Задание 3. 1000 нормально распределённых точек

n = 1000
mu = 0.1
variance = 5
sigma = np.sqrt(variance)

x = np.random.normal(loc=mu, scale=sigma, size=n)
y = np.random.normal(loc=mu, scale=sigma, size=n)

print(f"Количество точек: {n}")
print(f"Теоретическое среднее: {mu}")
print(f"Теоретическая дисперсия: {variance}")
print(f"Теоретическое стандартное отклонение: {sigma:.4f}")

print(f"\nВыборочное среднее X: {x.mean():.4f}")
print(f"Выборочная дисперсия X: {x.var():.4f}")
print(f"Выборочное среднее Y: {y.mean():.4f}")
print(f"Выборочная дисперсия Y: {y.var():.4f}")
```

```python
# Задание 3. Диаграмма рассеяния + тепловая карта

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Диаграмма рассеяния
axes[0].scatter(x, y, s=15, alpha=0.6)
axes[0].set_title("1000 точек")
axes[0].set_xlabel("x")
axes[0].set_ylabel("y")
axes[0].grid(True)

# Тепловая карта плотности
hist = axes[1].hist2d(x, y, bins=35)
axes[1].set_title("Тепловая карта плотности")
axes[1].set_xlabel("x")
axes[1].set_ylabel("y")
fig.colorbar(hist[3], ax=axes[1], label="Количество точек")

plt.tight_layout()
plt.show()
```

```python
# Задание 5. Загрузка датасета Rice Type Classification

import kagglehub

dataset_path = kagglehub.dataset_download(
    "mssmartypants/rice-type-classification"
)

print("Датасет загружен в:", dataset_path)

# Ищем CSV-файлы
csv_files = glob.glob(os.path.join(dataset_path, "**", "*.csv"), recursive=True)

print("\nНайденные CSV-файлы:")
for f in csv_files:
    print(" -", f)

if not csv_files:
    raise FileNotFoundError(
        "CSV-файл не найден. Скачайте датасет с Kaggle и загрузите CSV в Colab."
    )

# Если файлов несколько, берём первый и показываем список.
csv_path = csv_files[0]
df = pd.read_csv(csv_path)

print("\nИспользуется файл:", csv_path)
print("Размер таблицы:", df.shape)
display(df.head())
```

```python
# Задание 5. Информация о таблице

print("Названия столбцов:")
print(df.columns.tolist())

print("\nТипы данных:")
print(df.dtypes)

print("\nПропуски:")
print(df.isna().sum())

print("\nКоличество уникальных значений:")
print(df.nunique())
```

```python
# Задание 5. Поиск целевого столбца Class

# В датасете из задания целевой признак называется Class.
target_col = None

for col in df.columns:
    if str(col).strip().lower() == "class":
        target_col = col
        break

if target_col is None:
    raise ValueError(
        "Столбец Class не найден. Проверьте названия столбцов выше."
    )

print("Целевой признак:", target_col)
print("\nРаспределение классов:")
print(df[target_col].value_counts())
```

### 5.1. Гистограммы количественных признаков

```python
# Количественные признаки

numeric_cols = df.select_dtypes(include=np.number).columns.tolist()

# Id не является содержательным признаком для анализа,
# поэтому исключаем его из графиков, если он присутствует.
numeric_feature_cols = [
    col for col in numeric_cols
    if str(col).strip().lower() not in ["id", "index"]
]

print("Количественные признаки:")
print(numeric_feature_cols)

ncols = 3
nrows = int(np.ceil(len(numeric_feature_cols) / ncols))

fig, axes = plt.subplots(nrows, ncols, figsize=(15, 4 * nrows))
axes = np.array(axes).reshape(-1)

for ax, col in zip(axes, numeric_feature_cols):
    sns.histplot(data=df, x=col, kde=True, ax=ax)
    ax.set_title(f"Распределение: {col}")

for ax in axes[len(numeric_feature_cols):]:
    ax.remove()

plt.tight_layout()
plt.show()
```

### 5.2. Категориальный признак и распределение классов

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x=target_col)
plt.title("Распределение классов")
plt.xlabel("Класс")
plt.ylabel("Количество объектов")
plt.xticks(rotation=30)
plt.show()
```

### 5.3. Pairplot

Для pairplot используются числовые признаки и цветовая маркировка по классу.
Если признаков много, график может быть большим, но он соответствует задаче визуального анализа.

```python
# Pairplot
pairplot_cols = numeric_feature_cols + [target_col]

sns.pairplot(
    df[pairplot_cols].dropna(),
    hue=target_col,
    corner=True,
    diag_kind="hist"
)

plt.show()
```

### 5.4. Корреляционная матрица

```python
corr = df[numeric_feature_cols].corr()

plt.figure(figsize=(12, 9))
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)
plt.title("Корреляционная матрица количественных признаков")
plt.tight_layout()
plt.show()
```

### 5.5. Поиск наиболее эффективной пары признаков

Для объективного выбора пары используется однофакторный критерий ANOVA (`f_classif`) из `sklearn`.

Для каждой пары признаков рассчитывается разделяющая способность пары: сумма стандартизованных F-статистик двух признаков. Затем выбирается пара с максимальным значением.

Это является вычислительным способом дополнить визуальный анализ pairplot и корреляционной матрицы.

```python
# Подготовка данных для оценки признаков

analysis_df = df[numeric_feature_cols + [target_col]].dropna().copy()

X = analysis_df[numeric_feature_cols]
y = analysis_df[target_col]

f_values, p_values = f_classif(X, y)

feature_scores = pd.DataFrame({
    "Признак": numeric_feature_cols,
    "F_score": f_values,
    "p_value": p_values
}).sort_values("F_score", ascending=False)

print("Оценка признаков по ANOVA F-test:")
display(feature_scores)
```

```python
# Выбор лучшей пары по суммарному стандартизованному F-score

score_series = feature_scores.set_index("Признак")["F_score"]

# Нормируем F-score, чтобы сумма двух признаков была удобной для сравнения.
score_min = score_series.min()
score_max = score_series.max()

if score_max > score_min:
    normalized_scores = (score_series - score_min) / (score_max - score_min)
else:
    normalized_scores = score_series * 0

best_pair = None
best_pair_score = -np.inf

for i in range(len(numeric_feature_cols)):
    for j in range(i + 1, len(numeric_feature_cols)):
        f1 = numeric_feature_cols[i]
        f2 = numeric_feature_cols[j]
        pair_score = normalized_scores[f1] + normalized_scores[f2]

        if pair_score > best_pair_score:
            best_pair_score = pair_score
            best_pair = (f1, f2)

print("Наиболее эффективная пара признаков:")
print(best_pair)
print("Оценка пары:", best_pair_score)
```

```python
# 5.6. Визуализация распределения классов по лучшей паре

feature_x, feature_y = best_pair

plt.figure(figsize=(10, 7))
sns.scatterplot(
    data=analysis_df,
    x=feature_x,
    y=feature_y,
    hue=target_col,
    style=target_col,
    s=60,
    alpha=0.75
)

plt.title(
    f"Распределение классов по наиболее эффективной паре:\n"
    f"{feature_x} и {feature_y}"
)
plt.xlabel(feature_x)
plt.ylabel(feature_y)
plt.legend(title="Класс")
plt.grid(True)
plt.tight_layout()
plt.show()
```
